## Comparing the results of the edge prediction task using the pipeline, checking for bias

The dataframes with unbiased in the name, contain the results obtained from the edge_prediction_evaluation pipeline from GRAPE where the calculation of the embedding is left to the pipeline. => No bias in result

The dataframes without unbiased contain the results obtained from the edge_prediction_evaluation pipeline from GRAPE where the calculation of the embedding was done on the complete graph used for the task. => Bias in results

```python
import os
# change working directory so that it picks up the grapehelper library
os.chdir('/home/ftorgano/rna-kg-analysis')
print(os.getcwd())
```

```python
import pandas as pd
from scipy.stats import wilcoxon, kruskal
```

```python
edge_prediction_fullrnakg_dim10 = pd.read_csv('./RNA-KG_notebooks/Default_RNA-KG/reports/report7/csv/edge_prediction_fullrnakg_dim10.csv',index_col=0)
edge_prediction_fullrnakg_dim10_unbiased = pd.read_csv('./RNA-KG_notebooks/Default_RNA-KG/reports/report7/csv/edge_prediction_fullrnakg_dim10_unbiased.csv',index_col=0)
```

```python
edge_prediction_fullrnakg_dim10_unbiased
```

### Decision Trees

```python
results_10_tree_test = edge_prediction_fullrnakg_dim10[(edge_prediction_fullrnakg_dim10["evaluation_mode"]=="test") & (edge_prediction_fullrnakg_dim10["model_name"]=="Decision Tree Classifier")]["balanced_accuracy"]
results_10_tree_test_unbiased = edge_prediction_fullrnakg_dim10_unbiased[(edge_prediction_fullrnakg_dim10_unbiased["evaluation_mode"]=="test") & (edge_prediction_fullrnakg_dim10_unbiased["model_name"]=="Decision Tree Classifier")]["balanced_accuracy"]
```

```python
f"{results_10_tree_test_unbiased.mean():.2%}+-{results_10_tree_test_unbiased.std():.2%}"
```

```python
kruskal(results_10_tree_test,results_10_tree_test_unbiased)
```

```python
wilcoxon(results_10_tree_test,results_10_tree_test_unbiased)
```

```python
from scipy.stats import shapiro, ttest_ind, wilcoxon
# compare the results and check if the difference is meaningul
# use ttest if normal and wilcoxon if not
# calculate the paired difference between results
diff = results_10_tree_test - results_10_tree_test_unbiased
shapiro_test = shapiro(diff)
print(f'Shapiro normality test: t-value: {shapiro_test[0]:.2f}, p-value: {shapiro_test[1]:.5f}')
t_test = ttest_ind(results_10_tree_test,results_10_tree_test_unbiased)
print(f'Normal distribution: t-value: {t_test[0]:.2f}, p-value: {t_test[1]:.5f}')
wilcoxon_test = wilcoxon(results_10_tree_test,results_10_tree_test_unbiased)
print(f'Non-normal distribution: statistic: {wilcoxon_test[0]:.2f}, p-value: {wilcoxon_test[1]:.5f}')
```

### Random forest

```python
results_10_forest_test = edge_prediction_fullrnakg_dim10[(edge_prediction_fullrnakg_dim10["evaluation_mode"]=="test") & (edge_prediction_fullrnakg_dim10["model_name"]=="Random Forest Classifier")]["balanced_accuracy"]
results_10_forest_test_unbiased = edge_prediction_fullrnakg_dim10_unbiased[(edge_prediction_fullrnakg_dim10_unbiased["evaluation_mode"]=="test") & (edge_prediction_fullrnakg_dim10_unbiased["model_name"]=="Random Forest Classifier")]["balanced_accuracy"]
```

```python
f"{results_10_forest_test_unbiased.mean():.2%}+-{results_10_forest_test_unbiased.std():.2%}"
```

```python
results_10_forest = pd.DataFrame()
results_10_forest['Biased'] = results_10_forest_test
results_10_forest['Unbiased'] = results_10_forest_test_unbiased
results_10_forest
```

```python
kruskal(results_10_forest_test,results_10_forest_test_unbiased)
```

```python
wilcoxon(results_10_forest_test,results_10_forest_test_unbiased)
```

```python
from scipy.stats import shapiro, ttest_ind, wilcoxon
# compare the results and check if the difference is meaningul
# use ttest if normal and wilcoxon if not
# calculate the paired difference between results
diff = results_10_forest_test - results_10_forest_test_unbiased
shapiro_test = shapiro(diff)
print(f'Shapiro normality test: t-value: {shapiro_test[0]:.2f}, p-value: {shapiro_test[1]:.5f}')
t_test = ttest_ind(results_10_forest_test,results_10_forest_test_unbiased)
print(f'Normal distribution, t-test: t-value: {t_test[0]:.2f}, p-value: {t_test[1]:.5f}')
wilcoxon_test = wilcoxon(results_10_forest_test,results_10_forest_test_unbiased)
print(f'Non-normal distribution, Wilcoxon test: statistic: {wilcoxon_test[0]:.2f}, p-value: {wilcoxon_test[1]:.5f}')
```
